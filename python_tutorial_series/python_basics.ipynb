```python
def create_user_profile(name, email, phone, password, confirm_password, **kwargs):

    
    if not name.isalpha() or len(name) <= 5:
        return {
            "error": "Name must contain only alphabetic characters "
                     "and be longer than 5 letters."
        }

    
    if not phone.isdigit() or len(phone) != 10:
        return {
            "error": "Phone number must contain exactly 10 digits "
                     "with no alphabets or symbols."
        }


    if password != confirm_password:
        return {
            "error": "Password and Confirm Password do not match."
        }


    if not 8 <= len(password) <= 18:
        return {
            "error": "Password must be between 8 and 18 characters long."
        }


    has_upper = any(char.isupper() for char in password)
    has_lower = any(char.islower() for char in password)
    has_digit = any(char.isdigit() for char in password)
    has_special = any(not char.isalnum() for char in password)

    if not (has_upper and has_lower and has_digit and has_special):
        return {
            "error": "Password must contain at least one uppercase letter, "
                     "lowercase letter, digit, and special character."
        }


    if "gender" in kwargs:
        if isinstance(kwargs["gender"], str):
            kwargs["gender"] = kwargs["gender"].upper()


    user_data = {
        "name": name,
        "email": email,
        "phone": phone,
        "password": password
    }


    user_data.update(kwargs)


    return user_data


user = create_user_profile(
    "PrafullaTimilsina",
    "prafulla@example.com",
    "9812345678",
    "Pass@123",
    "Pass@123",
    gender="male",
    age=25,
    city="Kathmandu"
)

print(user)
```

```python
# Using super() to call parent's __init__
class Animal:
    def __init__(self, name):
        self.name = name
        print(f"Animal init: {self.name}")
                        
class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)        # call Animal.__init__
        self.breed = breed
        print(f"Dog init: {self.breed}")
        

d = Dog("Buddy", "Labrador")
print("Name :", d.name)
print("Breed:", d.breed)
```

```python
class Animal:
    def sound(self):
        print("Some generic sound") 

a = Animal()
a.sound()

class Dog(Animal):
    def sound(self):
        print("Woof!")

d = Dog()      
d.sound()     

class Cat(Animal):
    def sound(self):
        print("Meow!")

c= Cat()
c.sound()
```

```python
class Circle:
    def __init__(self, radius):
        self.radius = radius

    def __add__(self, other):
        return Circle(self.radius + other.radius)


c1 = Circle(5)
c2 = Circle(3)

c3 = c1 + c2

print("c1 + c2:", c3.radius)
```
