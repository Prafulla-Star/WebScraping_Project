# Store The Products into the DataBase

### 1) Loading the Json Data

```python
import json

with open(r"F:\Python_Project_1\all_products.json", "r", encoding='utf-8') as f:
    data=json.load(f)
```

### 2) Inspecting the data

```python
for key, values in data.items():
    print(f"| Category: {key} | Number of products: {len(values)} |")
```

### 3) Make the database

```python
import sqlite3
conn = sqlite3.connect("products.db")
```

### 4) Make the Cursor

```python
cursor = conn.cursor() # this cursor object will be used to execute SQL commands
```

### 5) Make the Table

```python
products_table = """CREATE TABLE IF NOT EXISTS products (
                id          INTEGER PRIMARY KEY AUTOINCREMENT,
                category TEXT   NOT NULL,
                name    TEXT   NOT NULL,
                rating  FLOAT,
                price   TEXT,
                link    TEXT,
                review_count TEXT
                [line hidden/redacted by a highlighted box]
                )"""

cursor.execute(products_table)
conn.commit()
```

### 6) Format the data

```python
# Flatten the JSON into a list of tuples for bulk insert

rows = []

for category, products in data.items():
    for product in products:
        rows.append((
            category,
            product.get('name'),
            product.get('rating'),
            product.get('price'),
            product.get('card_url'),
            product.get('review_count')
        ))

rows[0]
```

### 7) Insert Into DataBase

```python
insert_query = """INSERT INTO products (category, name, rating, price, link, review_count) VALUES (?, ?, ?, ?, ?, ?)"""

cursor.executemany(insert_query, rows)
conn.commit()
```

### 8) View the Data

```python
select_query = 'SELECT * FROM products LIMIT 2'
cursor.execute(select_query)
rows = cursor.fetchall()
rows
```

```python
for i in rows:
    id, item, name, rate, price, link, review = i
    print(id, item, name, rate, price, link, review)
    print("="*100)
```
